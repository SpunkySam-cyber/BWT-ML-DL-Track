### 1. Import Libraries

In this step, we import the necessary libraries for building and training our neural network:

- `numpy`: A library for numerical operations.
- `matplotlib.pyplot`: A library for data visualization.
- `tensorflow`: The core TensorFlow library for building and training neural networks.
- `tensorflow.keras.datasets`: A module containing various datasets, including MNIST.
- `tensorflow.keras.models`: A module for defining neural network models.
- `tensorflow.keras.layers`: A module containing various neural network layers.
- `tensorflow.keras.utils`: A utility module, including functions for converting class vectors to binary class matrices.

### 2. Load and Preprocess the Data

We load the MNIST dataset, which consists of 60,000 training images and 10,000 test images of handwritten digits:

- `mnist.load_data()`: Loads the MNIST dataset.
- `X_train` and `X_test`: Arrays containing the training and test images.
- `y_train` and `y_test`: Arrays containing the training and test labels.

Next, we normalize the pixel values of the images to the range 0-1 by dividing by 255.0:

- `X_train = X_train / 255.0`
- `X_test = X_test / 255.0`

We also convert the labels to one-hot encoded format using `to_categorical`:

- `y_train = to_categorical(y_train, 10)`
- `y_test = to_categorical(y_test, 10)`

### 3. Visualize the Data

To get a better understanding of the data, we visualize some sample images from the training set along with their labels. We use a 2x5 grid to display 10 images:

- `plt.figure(figsize=(10, 5))`: Sets the figure size.
- `plt.subplot(2, 5, i+1)`: Creates a subplot in a 2x5 grid.
- `plt.imshow(X_train[i], cmap='gray')`: Displays the image in grayscale.
- `plt.title(np.argmax(y_train[i]))`: Sets the title to the corresponding label.
- `plt.axis('off')`: Turns off the axis for a cleaner look.

### 4. Build the ANN Model

We define the architecture of our neural network using TensorFlow's Sequential API:

- `model = Sequential()`: Initializes the Sequential model.
- `model.add(Input(shape=(28, 28)))`: Defines the input shape of the data.
- `model.add(Flatten())`: Converts the 2D images into 1D vectors.
- `model.add(Dense(128, activation='relu'))`: Adds a dense layer with 128 neurons and ReLU activation.
- `model.add(Dense(64, activation='relu'))`: Adds another dense layer with 64 neurons and ReLU activation.
- `model.add(Dense(10, activation='softmax'))`: Adds the output layer with 10 neurons and softmax activation for classification.

### 5. Compile the Model

Next, we compile the model by specifying the loss function, optimizer, and evaluation metric:

- `loss='categorical_crossentropy'`: Specifies the loss function.
- `optimizer='adam'`: Specifies the optimizer.
- `metrics=['accuracy']`: Specifies the evaluation metric.

### 6. Train the Model

We train the model using the training data for 10 epochs with a batch size of 32, and we use 20% of the training data for validation:

- `epochs=10`: Specifies the number of epochs.
- `batch_size=32`: Specifies the batch size.
- `validation_split=0.2`: Specifies the fraction of training data to use for validation.

### 7. Evaluate the Model

After training, we evaluate the model on the test data to see how well it performs on unseen data:

- `test_loss, test_acc = model.evaluate(X_test, y_test)`: Evaluates the model on the test data.
- `print(f'Test accuracy: {test_acc:.4f}')`: Prints the test accuracy.

We plot the training and validation accuracy and loss to visualize the training process and see if the model is overfitting or underfitting:

- `plt.plot(history.history['loss'], label='Train Loss')`: Plots the training loss.
- `plt.plot(history.history['val_loss'], label='Validation Loss')`: Plots the validation loss.
- `plt.title('Model Loss')`: Sets the plot title.
- `plt.xlabel('Epoch')`: Sets the x-axis label.
- `plt.ylabel('Loss')`: Sets the y-axis label.
- `plt.legend()`: Adds a legend to the plot.

### 9. Make Predictions

Using the trained model, we make predictions on the test dataset:

- `predictions = model.predict(X_test)`: Predicts the labels for the test data.

### 10. Visualize Predictions

To better understand the model's performance, we visualize some sample predictions along with their true labels. This helps us see how accurately the model predicts the handwritten digits and identify any patterns or common errors:

- `plt.figure(figsize=(10, 5))`: Sets the figure size.
- `plt.subplot(2, 5, i+1)`: Creates a subplot in a 2x5 grid.
- `plt.imshow(X_test[i], cmap='gray')`: Displays the image in grayscale.
- `plt.title(f"Pred: {np.argmax(predictions[i])}\nTrue: {np.argmax(y_test[i])}")`: Sets the title to the predicted and true labels.
- `plt.axis('off')`: Turns off the axis for a cleaner look.


```python
# Import Libraries
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.layers import Input
from tensorflow.keras.utils import to_categorical

# Load and Preprocess the Data
(X_train, y_train), (X_test, y_test) = mnist.load_data()
X_train = X_train / 255.0
X_test = X_test / 255.0
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# Visualize the Data
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(np.argmax(y_train[i]))
    plt.axis('off')
plt.show()

# Build the ANN Model
model = Sequential()
model.add(Input(shape=(28, 28)))
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(10, activation='softmax'))

# Compile the Model
model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

# Train the Model
history = model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.2)

# Evaluate the Model
test_loss, test_acc = model.evaluate(X_test, y_test)
print(f'Test accuracy: {test_acc:.4f}')

# Plot Training History
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()

# Make Predictions
predictions = model.predict(X_test)
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_test[i], cmap='gray')
    plt.title(f"Pred: {np.argmax(predictions[i])}\nTrue: {np.argmax(y_test[i])}")
    plt.axis('off')
plt.show()
```
